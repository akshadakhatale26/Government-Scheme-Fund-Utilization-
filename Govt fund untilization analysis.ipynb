```python
# Visualize the number of records for each financial year using a bar chart.
df["Financial_Year"].value_counts().sort_index().plot(
kind = "bar",
title = "Records by Financial Year",
xlabel = "Financial Year",
ylabel = "Number of Records",
color = ["skyblue","#4C78A8","#3B82F6", "#1E3A8A"])

# Visualize the sanctioned, released, and utilized funds for each financial year.
year_fund = df.groupby("Financial_Year").agg(
                       sanctioned = ("Sanctioned_Amount_Lakhs","sum"),
                       released = ("Fund_Released_Lakhs","sum"),
                       utilized = ("Fund_Utilized_Lakhs","sum"))
year_fund.plot(
          kind = "bar",
          title = "sanctioned vs released vs utilized funds by financial year",
          xlabel = "Financial Year",
          ylabel = "Amount(Lakhs)",
          color = ["#BFDBFE", "#3B82F6", "#1E3A8A"])
```

```python
cols = [
    "Sanctioned_Amount_Lakhs",
    "Fund_Released_Lakhs",
    "Fund_Utilized_Lakhs",
    "Utilization_Percent",
    "Beneficiaries_Covered",
    "Physical_Target_Units",
    "Physical_Achievement_Units"
]

plt.figure(figsize=(15, 12))

for i, col in enumerate(cols):
    plt.subplot(3, 3, i + 1)
    sns.boxplot(y=df[col])
    plt.title(f"Boxplot of {col}")

plt.tight_layout()
plt.show()
```

```python
cols = [
    "Sanctioned_Amount_Lakhs",
    "Fund_Released_Lakhs",
    "Fund_Utilized_Lakhs",
    "Utilization_Percent",
    "Beneficiaries_Covered",
    "Physical_Target_Units",
    "Physical_Achievement_Units"
]

corr = df[cols].corr(method="pearson")

plt.figure(figsize=(11, 7))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Analysis of Government Fund Utilization")
plt.tight_layout()
plt.show()
```

### Conclusion:

The Government Fund Utilization Analysis examined fund allocation, release, utilization, physical achievement, and beneficiary coverage across states, departments, schemes, and financial years. The results show total sanctioned funds of ₹4,000.74 crore, recorded releases of ₹2,979.51 crore, and recorded utilization of ₹2,019.35 crore. This represents an overall utilization ratio of approximately 67.78% based on the notebook's imputed data.

West Bengal recorded the highest total fund release and utilization among the listed states, while Gujarat recorded the highest beneficiary coverage and physical achievement. The Ministry of Rural Development had the highest departmental funding and utilization totals. However, the overall fund utilization rate declined from 68.89% in 2021–22 to 66.99% in 2024–25, indicating that utilization did not increase proportionally with funding releases.

To improve performance, government departments should strengthen financial planning, monitor utilization gaps at the state and scheme levels, improve implementation and procurement timelines, and track physical outcomes alongside financial expenditure. Timely audit follow-up and better data validation are also important for reliable monitoring.

Overall, the analysis demonstrates how data analytics can help identify funding gaps, compare implementation performance, and support more informed decisions about government fund management. The findings should be validated against original records because missing-value imputation may affect the reported totals and performance indicators.
